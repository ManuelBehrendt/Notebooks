## The thumbnail

Your recipe gets a card on the [gallery page](https://manuelbehrendt.github.io/Mera.jl/stable/gallery/),
and the card needs one picture: the one that says best what this recipe makes. It is also the picture
at the top of this notebook, so there is only one file to produce.

Mera does the sizing, so there is nothing to download. Save the figure you like most, at whatever
size it comes out, then either call `makethumb` here:

```julia
makethumb("media/your_figure.png", "media/your_recipe_thumb.png")
```

or run the gallery wrapper once, which finds your `media/` and names the file after your notebook:

```bash
julia make_thumb.jl NNN_your_recipe
```

Either way you get an 800 by 450 PNG, the size every card uses. The whole picture is kept and padded
to shape, so a wide multi-panel figure keeps its outer panels. A gif works too. A movie does not:
save a frame instead, which is sharper anyway.
